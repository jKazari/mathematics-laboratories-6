```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
riding = [[17, 18, 21, 22, 19, 21, 25, 22, 25, 24],[3, 6, 3.5, 4, 5, 6.3, 4.5, 5, 4.5, 4]]
swimming = [[17, 18, 20, 19, 22, 21, 23, 19, 21, 24],[8, 9, 7, 10, 7.5, 9, 8, 7, 8.5, 9]]
sailing = [[31, 28, 29, 36, 27, 32, 34, 35, 33, 39],[4, 6.3, 6, 3, 5, 7.5, 2, 5, 7, 4]]

plt.scatter(x=riding[0], y=riding[1], c='cyan', marker='v', label='riding')
plt.scatter(x=swimming[0], y=swimming[1], c='magenta', marker='s', label='swimming')
plt.scatter(x=sailing[0], y=sailing[1], c='yellow', marker='p', label='sailing')

plt.xlabel('Age')
plt.ylabel('Hours')
plt.title('Activities Scatter Graph')
plt.legend()
```

```python
labels = ('Python', 'Scala', 'C#', 'Java', 'PHP')
index = (1, 2, 3, 4, 5)
sizes = [45, 10, 15, 30, 22]

plt.barh(index, sizes, tick_label=labels, color=('red', 'green', 'blue', 'yellow', 'orange'))

plt.ylabel('Usage')
plt.xlabel('Programming Languages')
```

```python
x = range(0,10)
y = range(-5,5)

figure_1 = plt.figure()

sub_plot_2_2_1 = figure_1.add_subplot(221)
sub_plot_2_2_1.set(title='(1,2,2)')
sub_plot_2_2_1.plot(x,y,'rD:')
sub_plot_2_2_1.plot(x,[i**2 for i in y],'co-')

sub_plot_2_2_2 = figure_1.add_subplot(222)
sub_plot_2_2_2.set(title='(2,2,2)')
sub_plot_2_2_2.scatter([x,y],[y,x])
```
