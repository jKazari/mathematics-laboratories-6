```python
import matplotlib.pyplot as plt
import numpy as np
```

Zadanie 1

```python
browsers = ["Chrome", "Safari", "Edge", "Firefox", "Samsung Internet", "Other"]
market_share = [65.7, 18.6, 5.3, 2.8, 2.5, 5.1]

fig, ax = plt.subplots(figsize=(9, 7))

wedges, texts, autotexts = ax.pie(
    market_share,
    labels=browsers,
    autopct="%1.1f%%",
    pctdistance=0.75,
    startangle=0,
    wedgeprops={"edgecolor": "white", "linewidth": 1.2},
)

for autotext in autotexts:
    autotext.set_fontsize(8)
    autotext.set_fontweight("bold")

ax.set_title("Global Browser Market Share (2024)", fontsize=13, fontweight="bold", pad=20)
```

Zadanie 2

```python
labels = ('Python', 'Scala', 'C#', 'Java', 'PHP', 'Rust')
index = (1, 2, 3, 4, 5, 6)

web_usage = [20, 2,  5, 10, 14,  3]
data_science_usage = [15, 8,  5, 15,  2,  6]
games_usage = [10, 1,  5,  5,  4,  2]
embedded_usage = [ 2, 0,  1,  2,  0, 18]

bottom_1 = [w + d for w, d in zip(web_usage, data_science_usage)]
bottom_2 = [b + g for b, g in zip(bottom_1, games_usage)]

plt.bar(index, web_usage, tick_label=labels, label='web', bottom=None)
plt.bar(index, data_science_usage, tick_label=labels, label='data science', bottom=web_usage)
plt.bar(index, games_usage, tick_label=labels, label='games', bottom=bottom_1)
plt.bar(index, embedded_usage, tick_label=labels, label='embedded', bottom=bottom_2)

plt.ylabel('Usage')
plt.xlabel('Programming Languages')
plt.title('Programming Language Usage by Domain')
plt.legend()
```

Zadanie 3

```python
BAR_WIDTH = 0.15

teama_results = (60, 75, 56, 62, 58, 70)
teamb_results = (55, 68, 80, 73, 55, 66)
teamc_results = (72, 60, 65, 50, 78, 55)

index_teama = [1, 2, 3, 4, 5, 6]
index_teamb = [i + BAR_WIDTH for i in index_teama]
index_teamc = [i + BAR_WIDTH * 2 for i in index_teama]

ticks = [i + BAR_WIDTH for i in index_teama]

tick_labels = ('Lab 1', 'Lab 2', 'Lab 3', 'Lab 4', 'Lab 5', 'Lab 6')

plt.bar(index_teama, teama_results, BAR_WIDTH, color='b', label='Team A')
plt.bar(index_teamb, teamb_results, BAR_WIDTH, color='g', label='Team B')
plt.bar(index_teamc, teamc_results, BAR_WIDTH, color='r', label='Team C')

plt.xlabel('Labs')
plt.ylabel('Scores')
plt.title('Scores by Lab')
plt.xticks(ticks, tick_labels)
plt.legend()
```
