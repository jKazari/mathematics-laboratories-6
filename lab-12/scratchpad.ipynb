```python
import matplotlib.pyplot as pyplot
```

```python
pyplot.plot([1,2,3,4,55,6],'k+:')

pyplot.xlabel('Time',fontsize=12)
pyplot.ylabel('Height',fontsize=12)
pyplot.title('Time vs Height')


pyplot.show()
```
